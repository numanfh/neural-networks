````python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.preprocessing import MinMaxScaler, StandardScaler

x, y = make_regression(n_samples=10, n_features=5)

def min_max_normalize(x):
    x_min = x.min(axis=0)
    x_max = x.max(axis=0)
    return (x - x_min) / (x_max - x_min)

def standardize(x):
    mean = x.mean(axis=0)
    std = x.std(axis=0)
    return (x - mean) / std

x_minmax = min_max_normalize(x)
x_std = standardize(x)

x_minmax_sk = MinMaxScaler().fit_transform(x)
x_std_sk = StandardScaler().fit_transform(x)

print("MinMax equal:", np.allclose(x_minmax, x_minmax_sk))
print("Standardization equal:", np.allclose(x_std, x_std_sk))
````
